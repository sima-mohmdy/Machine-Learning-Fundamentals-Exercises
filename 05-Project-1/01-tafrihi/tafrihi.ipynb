```python
# Drop columns
train_data = train_data.drop(['CancelTime', 'BillID', 'TicketID', 'UserID', 'Cancel', 'HashPassportNumber_p', 'HashEmail'], axis=1)
train_data
```

```python
# Drop columns
test_data = test_data.drop(['CancelTime', 'BillID', 'TicketID', 'UserID', 'Cancel', 'HashPassportNumber_p', 'HashEmail'], axis=1)
test_data
```

```python
#fill null values of 'VehicleClass' column in train data
train_data['VehicleClass'] = train_data['VehicleClass'].fillna(train_data['VehicleClass'].mode()[0])
```

```python
#fill null values of 'VehicleType' column in train data
train_data['VehicleType'] = train_data['VehicleType'].fillna("unkown")
```

```python
#fill null values of 'VehicleClass' column in test data
test_data['VehicleClass'] = test_data['VehicleClass'].fillna(test_data['VehicleClass'].mode()[0])
```

```python
#fill null values of 'VehicleType' column in test data
test_data['VehicleType'] = test_data['VehicleType'].fillna("unkown")
```

#feature extraction

```python
train_data.dtypes
```

```python
# compute object to datetime (column of 'Created' & 'DepartureTime' to datetime)
#Convert the string to datetime

train_data['Created'] = pd.to_datetime(train_data['Created'])
train_data['DepartureTime'] = pd.to_datetime(train_data['DepartureTime'])

train_data['weekday'] = train_data['Created'].dt.dayofweek
train_data['Hour'] = train_data['Created'].dt.hour
train_data['month'] = train_data['Created'].dt.month

train_data['Departure_weekday'] = train_data['DepartureTime'].dt.dayofweek
train_data['Departure_hour'] = train_data['DepartureTime'].dt.hour
train_data['Departure_month'] = train_data['DepartureTime'].dt.month
```

```python
# compute object to datetime (column of 'Created' & 'DepartureTime' to datetime)
#Convert the string to datetime

test_data['Created'] = pd.to_datetime(test_data['Created'])
test_data['DepartureTime'] = pd.to_datetime(test_data['DepartureTime'])

test_data['weekday'] = test_data['Created'].dt.dayofweek
test_data['Hour'] = test_data['Created'].dt.hour
test_data['month'] = test_data['Created'].dt.month

test_data['Departure_weekday'] = test_data['DepartureTime'].dt.dayofweek
test_data['Departure_hour'] = test_data['DepartureTime'].dt.hour
test_data['Departure_month'] = test_data['DepartureTime'].dt.month
```

```python
from sklearn.model_selection import train_test_split

X = train_data.drop('TripReason', axis=1)
y = train_data['TripReason']

X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.25, random_state=1)
```

```python
train_data
```

```python
from sklearn.preprocessing import OrdinalEncoder

categorical_columns = X_train.select_dtypes(include=['object']).columns.tolist()
for col in categorical_columns:
  # Apply OrdinalEncoder with handle_unknown='use_encoded_value'
  ordinal_encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)

  # Transform both training and test data
  X_train[col] = ordinal_encoder.fit_transform(X_train[[col]])
  X_val[col] = ordinal_encoder.transform(X_val[[col]])
  test_data[col] = ordinal_encoder.transform(test_data[[col]])

```

```python
X_train.head()
```

```python
X_val.head()
```

```python
test_data.head()
```

```python
bool_columns = X_train.select_dtypes(include=["bool"])

for col in bool_columns:
  X_train[col] = X_train[col].astype(int)
  X_val[col] = X_val[col].astype(int)
  test_data[col] = test_data[col].astype(int)
```

```python
X_train.head()
```

```python
X_val.head()
```

```python
test_data.head()
```

```python
X_train = X_train.drop(columns=['Created', 'DepartureTime'])
X_val = X_val.drop(columns=['Created', 'DepartureTime'])
test_data = test_data.drop(columns=['Created', 'DepartureTime'])
```

    <br>
    بر خلاف سایر تمرین‌های کالج، در این تمرین شما مجاز هستید از هر مدلی که آموخته‌اید استفاده کنید. به عبارت بهتر، هدف این سوال پیش‌بینی هرچه بهتر متغیر هدف مسئله است. انتظار ما این است که نحوه کار با کتابخانه‌های یادگیری ماشین را در حدی آموخته باشید که بتوانید با بهترین عملکرد، مسئله را حل کنید!

```python
# modeling
from sklearn.ensemble import RandomForestClassifier

classifier = RandomForestClassifier(n_estimators=100, random_state=42)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_val)

y_pred
```
